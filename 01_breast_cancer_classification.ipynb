```python
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np 


cancer = load_breast_cancer() 


X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
Y = pd.Series(cancer.target) 


X_train, X_test, Y_train, Y_test = train_test_split(
    X, 
    Y, 
    test_size=0.3, 
    random_state=42
)

# 5. Model Initialization (Optional but helpful next step)
# dtc = DecisionTreeClassifier() 
# dtc.fit(X_train, Y_train) 
# print(f"Training set shape: {X_train.shape}")
# print(f"Test set shape: {X_test.shape}")



# ... (Previous code for data loading and splitting)

# Train all three models
High_bias_model = DecisionTreeClassifier(max_depth=2)
High_bias_model.fit(X_train, Y_train)

medium_bias_model = DecisionTreeClassifier(max_depth=5)
medium_bias_model.fit(X_train, Y_train)

low_bias_model = DecisionTreeClassifier(max_depth=None)
low_bias_model.fit(X_train, Y_train)


# Corrected and completed list of Training Accuracies
train_accuracy = [
    High_bias_model.score(X_train, Y_train),  # max_depth=2
    medium_bias_model.score(X_train, Y_train), # max_depth=5
    low_bias_model.score(X_train, Y_train)    # max_depth=None
]

# Essential list of Testing Accuracies
test_accuracy = [
    High_bias_model.score(X_test, Y_test), 
    medium_bias_model.score(X_test, Y_test), 
    low_bias_model.score(X_test, Y_test)
]
```

```python
# Create a DataFrame using meaningful labels for the x-axis
df_plot = pd.DataFrame({
    # Use the max_depth or a descriptive label instead of the model object
    'Model Complexity': ['Max Depth 2', 'Max Depth 5', 'Max Depth None'], 
    'train_accuracy': train_accuracy,
    'test_accuracy': test_accuracy
})

# Plot using the descriptive labels
plt.plot(df_plot['Model Complexity'], df_plot['train_accuracy'], marker='o', label='Train', color='blue')
plt.plot(df_plot['Model Complexity'], df_plot['test_accuracy'], marker='o', label='Test', color='orange')
plt.title('Accuracy vs. Model Complexity')
plt.xlabel('Model Complexity')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(axis='y', linestyle='--')
plt.show()
```
